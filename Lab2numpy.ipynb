```python
import os
import numpy as np

filename = 'big.dat'
shape = (50000, 200)
dtype = np.float32

fp = np.memmap(filename, dtype=dtype, mode='w+', shape=shape)

batch_size = 5000
rng = np.random.default_rng(42)

for start in range(0, shape[0], batch_size):
    end = min(start + batch_size, shape[0])
    fp[start:end] = rng.random((end - start, shape[1]), dtype=dtype)

fp.flush()
del fp

expected_bytes = shape[0] * shape[1] * np.dtype(dtype).itemsize
actual_bytes = os.path.getsize(filename)

print(f"Ожидаемый размер: {expected_bytes} байт")
print(f"Фактический размер: {actual_bytes} байт")
```

```python
import numpy as np

mmap_arr = np.memmap('big.dat', dtype=np.float32, mode='r', shape=(50000, 200))

batch_size = 5000
global_min = np.inf
global_max = -np.inf
total_sum = 0.0
total_count = 0

for start in range(0, mmap_arr.shape[0], batch_size):
    end = min(start + batch_size, mmap_arr.shape[0])
    batch = mmap_arr[start:end]

    b_min = batch.min()
    b_max = batch.max()
    if b_min < global_min:
        global_min = b_min
    if b_max > global_max:
        global_max = b_max

    total_sum += batch.sum(dtype=np.float64)
    total_count += batch.size

global_mean = total_sum / total_count

print(f"Минимум:  {global_min:.6f}")
print(f"Максимум:  {global_max:.6f}")
print(f"Среднее: {global_mean:.6f}")

del mmap_arr
```

```python
import time
import numpy as np

N = 3000

arr_c = np.ones((N, N), dtype=np.float64, order='C')
arr_f = np.ones((N, N), dtype=np.float64, order='F')

def sum_by_rows(arr):
    total = 0.0
    rows, cols = arr.shape
    for i in range(rows):
        for j in range(cols):
            total += arr[i, j]
    return total

def sum_by_cols(arr):
    total = 0.0
    rows, cols = arr.shape
    for j in range(cols):
        for i in range(rows):
            total += arr[i, j]
    return total

def benchmark(func, arr, name):
    t0 = time.perf_counter()
    res = func(arr)
    elapsed = time.perf_counter() - t0
    print(f"{name:<32}: {elapsed:.4f} сек (сумма = {res:.0f})")
    return elapsed

print("C-order (построчно в памяти)")
t_c_rows = benchmark(sum_by_rows, arr_c, "Обход по строкам (i -> j)")
t_c_cols = benchmark(sum_by_cols, arr_c, "Обход по столбцам (j -> i)")

print("\nF-order (постолбцово в памяти)")
t_f_rows = benchmark(sum_by_rows, arr_f, "Обход по строкам (i -> j)")
t_f_cols = benchmark(sum_by_cols, arr_f, "Обход по столбцам (j -> i)")
```

```python
import time
import numpy as np

N = 10000000
rng = np.random.default_rng(42)
a = rng.uniform(1.0, 10.0, size=N)
b = rng.uniform(0.1, 0.9, size=N)

a_list = a.tolist()
b_list = b.tolist()

t0 = time.perf_counter()
res_list = [(x + y) / (x - y + 1.0) for x, y in zip(a_list, b_list)]
t_python = time.perf_counter() - t0

t0 = time.perf_counter()
res_np = (a + b) / (a - b + 1.0)
t_np_expr = time.perf_counter() - t0

out_buf = np.empty_like(a)
tmp_buf = np.empty_like(a)

t0 = time.perf_counter()
np.add(a, b, out=out_buf)
np.subtract(a, b, out=tmp_buf)
np.add(tmp_buf, 1.0, out=tmp_buf)
np.divide(out_buf, tmp_buf, out=out_buf)
t_np_out = time.perf_counter() - t0

assert np.allclose(res_np, out_buf)
print(f"1. Python циклы: {t_python:.4f} сек")
print(f"2. NumPy в одну строку:       {t_np_expr:.4f} сек")
print(f"3. NumPy с out= буферами:     {t_np_out:.4f} сек")
```

```python
import numpy as np
from scipy import sparse

rng = np.random.default_rng(42)

N_entries = 10_000
shape = (5000, 5000)

row = rng.integers(0, shape[0], size=N_entries).tolist()
col = rng.integers(0, shape[1], size=N_entries).tolist()
data = rng.normal(size=N_entries).tolist()

coo = sparse.coo_matrix((data, (row, col)), shape=shape)

csr = coo.tocsr()
csr.eliminate_zeros()

expected_entries = {}
for r, c, val in zip(row, col, data):
    expected_entries[(r, c)] = expected_entries.get((r, c), 0.0) + val

expected_nnz = sum(1 for val in expected_entries.values() if not np.isclose(val, 0.0, atol=1e-15))

print(f"Исходная длина списков: {len(data)}")
print(f"Уникальных пар (row, col): {len(expected_entries)}")
print(f"csr.nnz: {csr.nnz}, эталонный nnz: {expected_nnz}")
```

```python
import time
import numpy as np
from scipy import sparse

n = 4000
density = 0.005
rng = np.random.default_rng(42)

A_csr = sparse.random(n, n, density=density, format='csr', dtype=np.float64, random_state=rng)
A_dense = A_csr.toarray()
x = rng.standard_normal(n)

dense_bytes = A_dense.nbytes
csr_bytes = A_csr.data.nbytes + A_csr.indices.nbytes + A_csr.indptr.nbytes

print(f"Память Dense: {dense_bytes / 1024**2:.2f} МБ ({dense_bytes:,} байт)")
print(f"Память CSR:   {csr_bytes / 1024**2:.2f} МБ ({csr_bytes:,} байт)")
print(f"Экономия памяти: в {dense_bytes / csr_bytes:.1f} раз\n")

repeats = 200

_ = A_dense @ x
t0 = time.perf_counter()
for _ in range(repeats):
    y_dense = A_dense @ x
t_dense = (time.perf_counter() - t0) / repeats

_ = A_csr @ x
t0 = time.perf_counter()
for _ in range(repeats):
    y_csr = A_csr @ x
t_csr = (time.perf_counter() - t0) / repeats

assert np.allclose(y_dense, y_csr)
print(f"Время Dense (A @ x): {t_dense * 1e3:.4f} мс")
print(f"Время CSR   (A @ x): {t_csr * 1e3:.4f} мс")
print(f"CSR быстрее Dense в {t_dense / t_csr:.1f} раз")
```

```python
import time
import numpy as np
from scipy import sparse

n = 2000
densities = [0.001, 0.005, 0.01, 0.05, 0.1, 0.2, 0.5, 1.0]
repeats = 300
rng = np.random.default_rng(42)
x = rng.standard_normal(n)

print(f"{'Плотность':>10} | {'Dense (мс)':>12} | {'CSR (мс)':>12} | {'Отношение (Dense/CSR)':>22}")
print("-" * 65)

crossover_density = None

for d in densities:
    A_csr = sparse.random(n, n, density=d, format='csr', dtype=np.float64, random_state=rng)
    A_dense = A_csr.toarray()
    
    _ = A_dense @ x
    t0 = time.perf_counter()
    for _ in range(repeats):
        _ = A_dense @ x
    t_dense = (time.perf_counter() - t0) / repeats * 1e3

    _ = A_csr @ x
    t0 = time.perf_counter()
    for _ in range(repeats):
        _ = A_csr @ x
    t_csr = (time.perf_counter() - t0) / repeats * 1e3
    
    ratio = t_dense / t_csr
    print(f"{d:>10.3f} | {t_dense:>12.4f} | {t_csr:>12.4f} | {ratio:>21.2f}x")
    
    if crossover_density is None and t_csr > t_dense:
        crossover_density = d

print("-" * 65)
if crossover_density:
    print(f"Точка перелома: начиная с плотности ~{crossover_density} CSR проигрывает Dense.")
```
